## Why functions?

Without a function, the same calculation gets copy-pasted -- and copy-pasted code is where typos hide.

```python
# Cluttered: the same rectangle-area formula, repeated
area1 = 4 * 3
area2 = 6 * 2
area3 = 5 * 5
```

## Defining and calling a function

`def` names a function and starts its indented block of statements. Nothing runs until the function is *called*.

```python
def print_store_banner():
    print("Welcome to Prof. T's Bookstore")
```

```python
# still nothing happened above -- now call it
print_store_banner()
```

## The return statement

`return` sends a value back to the caller. A function with no `return` sends back `None`.

```python
def compute_double(num):
    return num * 2
```

```python
doubled = compute_double(21)
print(f"Double is {doubled}")
```

## Parameters and arguments

A *parameter* is the input named in the `def` line. An *argument* is the actual value passed in when the function is called.

```python
def convert_minutes_to_hours(total_minutes):
    return total_minutes / 60.0
```

```python
convert_minutes_to_hours(210.0)
```

## Multiple parameters

Arguments are matched to parameters in order: first to first, second to second, etc.

```python
def calc_rectangle_area(width, height):
    return width * height
```

```python
print(f"4 x 3 room: {calc_rectangle_area(4, 3)} sq ft")
print(f"6 x 2 room: {calc_rectangle_area(6, 2)} sq ft")
```

## Hierarchical (nested) function calls

A function's body can call *another* function -- here, `calc_paint_cost()` calls `calc_rectangle_area()` to do part of its job.

```python
def calc_paint_cost(width, height, price_per_sqft):
    area = calc_rectangle_area(width, height)
    return area * price_per_sqft
```

```python
print(f"Cost to paint: ${calc_paint_cost(10, 8, 1.50):.2f}")
```
