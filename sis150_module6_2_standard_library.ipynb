## What is the standard library?

Python ships with a large collection of modules ready to `import` --
no installation needed. Before writing your own helper code, it's
worth checking whether the standard library already has it. A few
commonly used modules:

| Module     | What it's for                                  |
|------------|-------------------------------------------------|
| `datetime` | Dates, times, and the arithmetic between them   |
| `random`   | Random numbers, random choices, shuffling       |
| `math`     | Mathematical functions (`sqrt`, `ceil`, `floor`, `pi`) |
| `os`       | Talking to the operating system -- files, paths |

This notebook tries out four of them.


## `datetime` -- date arithmetic

`datetime.date.today()` gives today's date as a `date` object.
`datetime.timedelta` represents a *span* of time (a number of days,
here) and can be added to or subtracted from a date.


```python
from datetime import date, timedelta

today = date.today()  # today is a date object, not a string
abbey_road_release = date(1969, 9, 26)  # real release date, UK

# subtracting two date objects gives back a timedelta -- an object that
# represents the *span* between them. .days pulls out just the whole
# number of days as an int
days_since_release = today - abbey_road_release
print(f"Today: {today}")
print(f"Abbey Road was released: {abbey_road_release}")
print(f"That was {days_since_release.days:,} days ago.")  # the , adds thousands separators
```

```python
# timedelta(days=365) is a 365-day span; adding it to a date gives back
# a new date object -- the original abbey_road_release is unchanged
one_year_later = abbey_road_release + timedelta(days=365)
print(f"One year after release: {one_year_later}")
```

## `random` -- shuffling and random choices

`random.sample(sequence, k)` picks `k` items without repeats -- handy
for building a shuffled play queue from a track list. `random.choice`
picks a single item; calling it again can repeat a track, since it
doesn't remove anything.


```python
import random

abbey_road_side_b = [
    "Here Comes The Sun",
    "Because",
    "You Never Give Me Your Money",
    "Sun King",
    "Mean Mr Mustard",
    "Polythene Pam",
    "She Came In Through The Bathroom Window",
    "Golden Slumbers",
    "Carry That Weight",
    "The End",
]

# random.sample() returns a NEW list, same items, shuffled order.
# k=len(...) samples every item exactly once -- a full shuffle, no repeats
shuffled_queue = random.sample(abbey_road_side_b, k=len(abbey_road_side_b))
print("Shuffled play queue:")
# enumerate() pairs each item with a running count; start=1 makes that
# count begin at 1 instead of Python's usual 0
for position, track_name in enumerate(shuffled_queue, start=1):
    print(f"  {position}. {track_name}")
```

```python
# random.choice can repeat -- it doesn't remove the item it picks
print("Random pick:", random.choice(abbey_road_side_b))
print("Random pick:", random.choice(abbey_road_side_b))
```

## `math` -- ceil, floor, and time formatting

`data/abbey_road_tracks.csv` (built in Module 6.1) has each track's
duration in minutes as a float, e.g. `4.33`. `math.ceil()` and
`math.floor()` round up/down to the nearest whole number.


```python
import math

with open("data/abbey_road_tracks.csv") as tracks_file:
    lines = tracks_file.readlines()

data_lines = lines[1:]  # skip header

# for each line: strip the \n, split on ",", take field index 3
# (track_duration_min) as a str, convert it to float, then sum() adds
# up all 17 of those floats into one total
total_minutes = sum(float(line.strip().split(",")[3]) for line in data_lines)

print(f"Total Abbey Road runtime: {total_minutes:.2f} minutes")
print(f"Rounded up (math.ceil):   {math.ceil(total_minutes)} minutes")
print(f"Rounded down (math.floor): {math.floor(total_minutes)} minutes")
```

```python
# Split total_minutes (a float, e.g. 47.40) into whole minutes + seconds
whole_minutes = math.floor(total_minutes)  # the whole-number part
leftover_seconds = round((total_minutes - whole_minutes) * 60)  # the fractional part, in seconds
print(f"That's {whole_minutes} min {leftover_seconds} sec.")
```

## `os` -- files and paths

The `os` module answers questions about the file system: what's the
current working directory, what files exist, does this path exist?
`os.path.exists()` is a safe way to check before opening a file
(Module 6.4 covers what happens if you *don't* check).


```python
import os

print("Current working directory:", os.getcwd())  # returns a str
print("Files in data/:", os.listdir("data"))  # returns a list of str filenames
```

```python
# os.path.join() builds a path string using the right separator for the
# current OS ("/" on macOS and Linux, "\" on Windows) -- more portable
# than typing "data/abbey_road_tracks.csv" by hand
tracks_path = os.path.join("data", "abbey_road_tracks.csv")
missing_path = os.path.join("data", "does_not_exist.csv")

print(tracks_path, "exists:", os.path.exists(tracks_path))  # os.path.exists() returns a bool
print(missing_path, "exists:", os.path.exists(missing_path))
```

  time you can add or subtract.
- `random.sample()` shuffles without repeats; `random.choice()` can
  repeat, since nothing is removed from the source list.
- `math` has the functions you'd expect from a calculator: `ceil`,
  `floor`, `sqrt`, `pi`, and more.

- Before writing a helper function from scratch, it's worth checking

  https://docs.python.org/3/library/
