## List basics

A list is a **container** -- it groups related values together. A list is **mutable** -- it can grow, shrink, and change after it's created. A list is also a **sequence** -- its elements keep a left-to-right order.

```python
watchlist = ["Inception", "Arrival", "Dune"]
watchlist
```

```python
type(watchlist)
```

A list can also be built from another iterable using the built-in `list()` function.

```python
list("Dune")
```

## Accessing list elements

Each element has a zero-based index. `watchlist[0]` is the first element, `watchlist[-1]` is the last.

```python
watchlist[0]
```

```python
watchlist[2]
```

```python
watchlist[-1]
```

```python
# an index can be an expression, as long as it evaluates to an int
i = 1
watchlist[i]
```

## Common list operations

Slicing returns a *new* list. Concatenation (`+`) also returns a *new* list. Assigning to an index changes the list *in place*.

```python
watchlist[0:2]  # slice -- a new list
```

```python
watchlist + ["Interstellar"]  # concatenation -- a new list
```

```python
watchlist[1] = "Oppenheimer"  # in-place: changes watchlist itself
watchlist
```

```python
del watchlist[0]  # in-place: removes an element by index
watchlist
```

## In-place modification and shared references

Careful: if two variables point to the *same* list, an in-place change through one variable shows up through the other too.

```python
my_watchlist = ["Inception", "Arrival", "Dune"]
your_watchlist = my_watchlist  # SAME list, not a copy

my_watchlist[0] = "Tenet"
print(my_watchlist)
print(your_watchlist)  # changed too!
```

To make an independent copy, slice the whole list with `[:]`.

```python
my_watchlist = ["Inception", "Arrival", "Dune"]
your_watchlist = my_watchlist[:]  # an actual COPY

my_watchlist[0] = "Tenet"
print(my_watchlist)
print(your_watchlist)  # unaffected
```
