## Writing a text file

The built-in `open()` function returns a **file object**. Pass a
second argument, `'w'`, to open the file for **writing** -- if the
file doesn't exist yet, it's created; if it does exist, its old
contents are overwritten. The `write()` method writes a string to the
file. `close()` releases the file when you're done with it.


```python
# adjacent string literals are concatenated automatically -- no + needed.
# the parentheses here just let the expression span multiple lines;
# they do NOT make this a tuple (that would require a comma).
session_notes = (
    "Abbey Road recording sessions\n"
    "EMI Studios, London -- February through August 1969\n"
    "Producer: George Martin\n"
    "Engineer: Geoff Emerick\n"
)

notes_file = open("data/abbey_road_session_notes.txt", "w")
notes_file.write(session_notes)
notes_file.close()

print("Wrote data/abbey_road_session_notes.txt")
```

## Reading a text file

Open the same file in `'r'` mode (the default -- `open(fname)` alone
also works). `read()` returns the entire file as one string.


```python
notes_file = open("data/abbey_road_session_notes.txt", "r")
contents = notes_file.read()  # contents is one big str -- the whole file, in one piece
notes_file.close()

print(contents)
```

`readlines()` instead returns a **list of strings**, one per line
(each line keeps its trailing `\n`).


```python
notes_file = open("data/abbey_road_session_notes.txt")
lines = notes_file.readlines()  # lines is a list of str -- one str per line, each still ending in \n
notes_file.close()

lines
```

Iterating over a file's lines is common enough that a file object
supports `for line in file_object:` directly -- no need to call
`readlines()` first.


```python
notes_file = open("data/abbey_road_session_notes.txt")

for line in notes_file:
    print(line, end="")  # line already ends in \n, so end="" avoids an extra blank line

notes_file.close()
```

## The `with` statement

Forgetting to call `close()` -- or skipping it because an exception
happened first -- leaves a file handle open. The `with` statement
opens the file, hands it to you under whatever name you choose after
`as` (here, `notes_file`), and **closes it automatically** when the
block ends, even if an error occurs partway through. This is the
recommended, modern way to work with files in Python.


```python
with open("data/abbey_road_session_notes.txt") as notes_file:
    contents = notes_file.read()

# notes_file is already closed here -- no explicit close() needed
print(contents)
```

## File modes: `'r'`, `'w'`, and `'a'`

| Mode | Meaning                                    |
|------|---------------------------------------------|
| `'r'`| Read (default). Errors if the file is missing. |
| `'w'`| Write. Creates the file if missing, **overwrites** if it exists. |
| `'a'`| Append. Creates the file if missing, adds to the **end** if it exists. |

`'w'` mode is destructive -- reopening a file with `'w'` erases what
was there before. Use `'a'` to add on to an existing file instead.


```python
with open("data/abbey_road_session_notes.txt", "a") as notes_file:
    notes_file.write("Note: the Side B medley was largely tracked as one\n")
    notes_file.write("continuous block, stitched together in the mix.\n")

with open("data/abbey_road_session_notes.txt") as notes_file:
    print(notes_file.read())
```

```python
# 'w' mode overwrites -- everything appended above is gone after this
with open("data/abbey_road_session_notes.txt", "w") as notes_file:
    notes_file.write("Abbey Road recording sessions\n")
    notes_file.write("EMI Studios, London -- February through August 1969\n")

with open("data/abbey_road_session_notes.txt") as notes_file:
    print(notes_file.read())
```

## Processing data from a file

A common task: read a data file and compute something from it. Below,
`data/abbey_road_tracks.csv` is read one line at a time, by hand --
no `csv` module, no pandas yet (that comes in Module 6.3) -- to
compute the album's average Spotify track-popularity score.

Each line is a comma-separated string, e.g.
`"7,Here Comes The Sun,82,3.09\n"`. `strip()` removes the trailing
newline, and `split(",")` breaks the line into a list of fields.


```python
with open("data/abbey_road_tracks.csv") as tracks_file:
    lines = tracks_file.readlines()  # lines: a list of str, one per CSV row (header included)

# lines[0] is the header row as one string, e.g.
# "track_number,track_name,track_popularity,track_duration_min\n"
# .strip() drops the trailing \n, then .split(",") breaks that one string
# into a list of column-name strings
header = lines[0].strip().split(",")
data_lines = lines[1:]  # skip the header row -- everything from index 1 onward

print("Columns:", header)
print("Track rows:", len(data_lines))  # len() counts the list's elements: 17 track rows
```

```python
total_popularity = 0

for line in data_lines:
    # this time, strip().split(",") breaks one *data* row (not the header)
    # into its 4 field strings, e.g. "7,Here Comes The Sun,82,3.09" ->
    # ["7", "Here Comes The Sun", "82", "3.09"]
    fields = line.strip().split(",")
    # fields[2] is the track_popularity column -- split() always returns
    # strings, so int() converts "77" (a str) into 77 (an int) so it can be
    # added to total_popularity
    track_popularity = int(fields[2])
    total_popularity += track_popularity

average_popularity = total_popularity / len(data_lines)  # len(data_lines) is 17 tracks
print(f"Average Abbey Road track popularity: {average_popularity:.1f}")
```

### Try it yourself

Using the same `data_lines` list, compute the **longest** track on
the album (the `track_duration_min` column, index 3) and print its
name (index 1).


```python
longest_name = None
longest_duration = 0.0

for line in data_lines:
    # third and last use of this pattern -- same idea, different fields
    # pulled out of the row this time
    fields = line.strip().split(",")
    track_name = fields[1]  # fields[1] is track_name -- stays a str, no conversion needed
    # fields[3] is track_duration_min -- also a plain str fresh out of
    # split(), e.g. "4.33". Converted with float(), not int(), because
    # duration has decimal places (unlike track_popularity above, which
    # is always a whole number, e.g. "77")
    track_duration_min = float(fields[3])
    if track_duration_min > longest_duration:
        longest_duration = track_duration_min
        longest_name = track_name

print(f"Longest track: {longest_name} ({longest_duration} min)")
```

## Let's review

- `open(fname, mode)` returns a file object; `'r'` reads, `'w'`
  writes/overwrites, `'a'` appends.
- `read()` gets the whole file as one string; `readlines()` gets a
  list of lines; `for line in f:` iterates lines directly.
- Prefer `with open(...) as f:` over manual `open()`/`close()` -- it
  closes the file for you automatically.
- Reading a CSV by hand with `split(",")` works for simple files, but
  gets fragile fast (quoted fields, embedded commas). Module 6.3
  brings in pandas to do this properly.
