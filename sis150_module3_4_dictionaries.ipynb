## Dictionaries store key-value pairs

Like an English dictionary matches a word to a definition, a Python `dict` matches a *key* to a *value*.

```python
contacts = {
    "Ana": "555-0132",
    "Bo": "555-0198",
    "Cy": "555-0177"
}
contacts
```

```python
contacts["Ana"]  # indexing a dict by key
```

A dict can also be created with the built-in `dict()` function.

```python
dict(Ana="555-0132", Bo="555-0198")
```

## Adding, modifying, and deleting entries

Dictionaries are mutable, so entries can change in place.

```python
# Adding a new entry
contacts["Dee"] = "555-0144"
contacts
```

```python
# Modifying an existing entry
contacts["Ana"] = "555-9999"
contacts
```

```python
# Deleting an entry
del contacts["Bo"]
contacts
```

## Testing whether a key exists

Use `in` *before* you look something up, to avoid an error.

```python
"Ana" in contacts
```

```python
"Bo" in contacts  # already deleted above
```

## A dict value can be anything -- even a list

Here, each key maps to a *list* of quiz scores.

```python
student_scores = {}
student_scores["Ana"] = [88, 92]
student_scores["Bo"] = [75, 81, 79]
student_scores
```

```python
# average score for Ana
sum(student_scores["Ana"]) / len(student_scores["Ana"])
```
