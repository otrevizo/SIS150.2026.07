## Defining a class

A class is a blueprint. It groups related data (attributes) together under one name.

```python
class Song:
    """A song in a digital music library."""
    def __init__(self):
        self.title = ""
        self.artist = ""
        self.duration_sec = 0
```

## Instantiation

Calling the class like a function creates a new *instance* -- an individual object built from the blueprint.

```python
song1 = Song()
song1.title = "Come Together"
song1.artist = "The Beatles"
song1.duration_sec = 259

print(song1.title, "--", song1.artist)
```

## Multiple instances

Each instance keeps its own copy of the data.

```python
song2 = Song()
song2.title = "Paint It Black"
song2.artist = "The Rolling Stones"
song2.duration_sec = 202

print(song1.title, song1.duration_sec)
print(song2.title, song2.duration_sec)
```

## Constructor parameters

Setting every attribute by hand is repetitive. Add parameters to `__init__` so we can set them at creation time.

```python
class Song:
    """A song in a digital music library."""
    def __init__(self, title, artist, duration_sec):
        self.title = title
        self.artist = artist
        self.duration_sec = duration_sec
```

```python
song1 = Song("With or Without You", "U2", 297)
song2 = Song("Air on the G String", "Bach", 300)

print(song1.title, song1.duration_sec)
print(song2.title, song2.duration_sec)
```

## Default parameter values

Give a parameter a default so it becomes optional at creation time.

```python
class Song:
    """A song in a digital music library."""
    def __init__(self, title, artist, duration_sec=180, genre="Pop"):
        self.title = title
        self.artist = artist
        self.duration_sec = duration_sec
        self.genre = genre
```

```python
# uses both defaults -- notice "Pop" doesn't really fit this one
song3 = Song("Take the A Train", "Oscar Peterson")
print(song3.title, song3.duration_sec, song3.genre)
```

```python
# override the defaults with the real values
song4 = Song("Take the A Train", "Oscar Peterson", duration_sec=195, genre="Jazz")
print(song4.title, song4.duration_sec, song4.genre)
```

## Instance methods

A function defined inside a class is a *method*. Every method's first parameter is `self` -- Python passes the calling instance automatically.

```python
class Song:
    """A song in a digital music library."""
    def __init__(self, title, artist, duration_sec=180, genre="Pop"):
        self.title = title
        self.artist = artist
        self.duration_sec = duration_sec
        self.genre = genre
        self.play_count = 0

    def play(self):
        self.play_count += 1
        print(f"Now playing: {self.title} by {self.artist}")

    def duration_display(self):
        minutes = self.duration_sec // 60
        seconds = self.duration_sec % 60
        return f"{minutes}:{seconds:02d}"
```

```python
song1 = Song("Eine kleine Nachtmusik", "Mozart", 366, "Classical")
song1.play()
song1.play()

print(f"Played {song1.play_count} times, duration {song1.duration_display()}")
```

## A common bug: forgetting `self`

If a method's first parameter isn't `self`, Python still passes the instance automatically when you call it -- and the call fails.

```python
class BuggySong:
    def __init__(self, title):
        self.title = title
    def show_title():          # forgot self!
        print(self.title)

try:
    b = BuggySong("Mystery Track")
    b.show_title()
except TypeError as e:
    print(f"TypeError: {e}")
```

## Class attributes

A class attribute is defined in the class body (not inside `__init__`) and is shared by every instance.

```python
class Song:
    """A song in a digital music library."""
    platform_name = "MyTunes"   # class attribute -- shared by all songs
    song_count = 0               # class attribute used as a running counter

    def __init__(self, title, artist, duration_sec=180, genre="Pop"):
        self.title = title
        self.artist = artist
        self.duration_sec = duration_sec
        self.genre = genre
        self.play_count = 0
        Song.song_count += 1     # every new Song increments the shared counter

    def play(self):
        self.play_count += 1
        print(f"Now playing: {self.title} by {self.artist}")

    def duration_display(self):
        minutes = self.duration_sec // 60
        seconds = self.duration_sec % 60
        return f"{minutes}:{seconds:02d}"
```

```python
song1 = Song("F\u00fcr Elise", "Beethoven", 180, "Classical")
song2 = Song("Summertime", "Ella Fitzgerald", 320, "Jazz")

print(song1.platform_name, song2.platform_name)   # same for both -- class namespace
print("Songs created so far:", Song.song_count)
```

## Instance attributes always win

If an instance sets its own attribute with the same name, Python finds the instance's value first -- the instance namespace is searched before the class namespace.

```python
song2.platform_name = "IndiePlayer"   # this creates a new instance attribute

print(song1.platform_name)   # still looks up the class attribute
print(song2.platform_name)   # instance attribute shadows the class attribute
```

## Changing a class attribute after instances already exist

Here's the surprising part: an instance doesn't get its own private
*copy* of a class attribute when it's created -- it looks the value up
**live**, in the class, every single time you ask for it. So if you
change the class attribute later, every instance that hasn't shadowed
it updates immediately too, even instances that already existed *before*
the change.

```python
# song1 never set its own platform_name (see the cell above), so it's
# still reading straight from the Song class -- live, not a snapshot
# frozen back when song1 was created.
Song.platform_name = "TuneWave"   # change the CLASS attribute itself

print(song1.platform_name)   # updates automatically -- song1 was never
                              # shadowed, so it re-reads the class again
print(song2.platform_name)   # unaffected -- song2 already has its OWN
                              # instance attribute (set in the cell
                              # above), and an instance attribute always
                              # wins over the class attribute
```

This is the exact same instance-namespace-then-class-namespace lookup
from the cell above -- `song1.platform_name` isn't found in `song1`'s
own namespace, so Python walks up and reads whatever is *currently* in
the class's namespace. It's not a copy taken at creation time -- it's a
live lookup, every time.

## Mini-challenge: a VinylRecord's value

Parallels the zyBooks *Car value* lab -- but instead of a car losing value, a collectible vinyl record tends to *gain* value over time.

```python
class VinylRecord:
    """A collectible vinyl record."""
    def __init__(self, title, purchase_price, purchase_year):
        self.title = title
        self.purchase_price = purchase_price
        self.purchase_year = purchase_year

    def current_value(self, current_year, annual_growth=0.08):
        age = current_year - self.purchase_year
        return self.purchase_price * (1 + annual_growth) ** age
```

```python
record1 = VinylRecord("Abbey Road", purchase_price=5.00, purchase_year=1970)
value_now = record1.current_value(current_year=2026)

print(f"{record1.title}: bought in {record1.purchase_year} for "
      f"${record1.purchase_price:.2f}, worth ${value_now:,.2f} in 2026")
```

`annual_growth` has a default (`0.08`), the same way a constructor
parameter can -- and it can be overridden the same way too, by passing
an argument when you *call* the method.

```python
# Same record, a more conservative growth assumption -- override the default
conservative_value = record1.current_value(current_year=2026, annual_growth=0.03)

print(f"At the default 8% growth: ${value_now:,.2f}")
print(f"At a more conservative 3% growth: ${conservative_value:,.2f}")
```

## Mini-challenge: a StreamingPlan's monthly cost

Parallels the zyBooks *Nutritional information* lab -- a constructor with defaults, plus a method that computes a weighted total from several attributes.

```python
class StreamingPlan:
    """A monthly music streaming plan."""
    def __init__(self, plan_name="Free Tier", hours_streamed=0.0,
                 downloads=0, offline_devices=0):
        self.plan_name = plan_name
        self.hours_streamed = hours_streamed
        self.downloads = downloads
        self.offline_devices = offline_devices

    def monthly_cost(self):
        # Free Tier is ad-supported -- it's free BECAUSE of the plan,
        # not because you happened to use less. Every other plan pays
        # per usage, same weighted formula as before.
        if self.plan_name == "Free Tier":
            return 0.00
        return (self.hours_streamed * 0.10
                + self.downloads * 0.05
                + self.offline_devices * 1.50)
```

```python
free_plan = StreamingPlan()
my_plan = StreamingPlan("Premium", hours_streamed=40.0, downloads=25, offline_devices=2)

print(f"{free_plan.plan_name}: ${free_plan.monthly_cost():.2f}/month")
print(f"{my_plan.plan_name}: ${my_plan.monthly_cost():.2f}/month")
```

Proof it's the *plan* that matters, not the usage numbers: give both
plans the exact same hours, downloads, and devices, and change only
`plan_name`.

```python
# Identical usage on both -- only plan_name differs
free_plan = StreamingPlan("Free Tier", hours_streamed=40.0, downloads=25, offline_devices=2)
my_plan = StreamingPlan("Premium", hours_streamed=40.0, downloads=25, offline_devices=2)

print(f"{free_plan.plan_name}: ${free_plan.monthly_cost():.2f}/month")   # $0.00, always
print(f"{my_plan.plan_name}: ${my_plan.monthly_cost():.2f}/month")       # billed per usage
```

## Mini-challenge: a RecordStore's inventory

Parallels the zyBooks *Vending machine* lab -- an object that starts with initial stock, then sells and restocks over a sequence of operations.

```python
class RecordStoreInventory:
    """Tracks vinyl record stock for one title at a record store."""
    def __init__(self, starting_stock=50):
        self.stock = starting_stock

    def sell(self, count):
        self.stock -= count

    def restock(self, count):
        self.stock += count

    def report(self):
        print(f"Inventory: {self.stock} records")
```

```python
# Abbey Road, continuing from the VinylRecord mini-challenge above
shop = RecordStoreInventory(starting_stock=30)
shop.sell(5)
shop.restock(2)
shop.sell(7)
shop.report()
```

## Let's review

- A **class** is a blueprint that groups attributes and methods.
- The **constructor** (`__init__`) sets up a new instance, optionally with default parameter values.
- **Instance methods** operate on one instance via `self`.
- A **class attribute** is shared by every instance; an **instance attribute** belongs to just one object -- and shadows a class attribute of the same name.
- Class attributes are looked up **live**, not copied at creation time -- change one after instances already exist, and every unshadowed instance reflects the new value immediately.
