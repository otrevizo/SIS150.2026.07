## Without exception handling

Suppose a "number of takes" field comes in as `"four"` instead of
`4`. `int("four")` raises a `ValueError`, and -- with no handling --
that error propagates up and crashes the program.

The cell below deliberately triggers that same `ValueError`, but
catches it just long enough to print the real traceback with
`traceback.print_exc()`, so you can see exactly what an *unhandled*
crash looks like without actually taking down this notebook's kernel.


```python
import traceback

try:
    number_of_takes = int("four")  # int() can't parse a word -- raises ValueError
except ValueError:
    # traceback.print_exc() prints the exact same message Python would show
    # if this weren't caught -- it's here only so you can see it safely.
    # Your own code normally won't call traceback.print_exc() itself.
    traceback.print_exc()
```

## The try/except construct

Code that might fail goes in a `try` block. If it raises an
exception, control jumps straight to the matching `except` block
instead of crashing -- and the program keeps going afterward.


```python
try:
    number_of_takes = int("four")
except ValueError:
    print("Could not read number of takes -- defaulting to 1.")
    number_of_takes = 1  # a fallback value, so the program has something usable

# unlike the traceback demo above, execution reaches this line either way --
# the program never crashed
print("number_of_takes =", number_of_takes)
```

## Common exception types

| Exception type     | Raised when...                                   |
|---------------------|--------------------------------------------------|
| `ValueError`        | A value has the right type but the wrong content (`int("four")`) |
| `KeyError`           | A dictionary key doesn't exist                    |
| `ZeroDivisionError`  | Dividing by zero (e.g., averaging an empty list) |
| `IndexError`         | A list/sequence index is out of bounds            |
| `EOFError`           | `input()` hits end-of-file with no data to read (not demoed here -- notebooks don't hit real EOF the way a terminal script does) |


### `KeyError`

Looking up a song in a dictionary by an exact title that doesn't
exist raises `KeyError`.


```python
song_to_album = {
    "Come Together": "Abbey Road",
    "Let It Be": "Let It Be",
    "Eleanor Rigby": "Revolver",
}

song_title = "Yesterday"  # not in the dict above

try:
    # dict[key] raises KeyError immediately if key is missing -- contrast
    # with dict.get(key), which would just return None instead
    album_name = song_to_album[song_title]
    print(f'"{song_title}" is on {album_name}.')
except KeyError:
    print(f'"{song_title}" is not in song_to_album.')
```

### `ZeroDivisionError`

Averaging an empty list of ratings divides by zero.


```python
fan_ratings = []  # no ratings submitted yet

try:
    # sum([]) is 0 and len([]) is 0 -- so this becomes 0 / 0, which raises
    # ZeroDivisionError (not a special case for empty lists specifically --
    # any division by zero triggers the same exception)
    average_rating = sum(fan_ratings) / len(fan_ratings)
    print(f"Average rating: {average_rating:.1f}")
except ZeroDivisionError:
    print("No ratings yet -- can't compute an average.")
```

### `IndexError`

Track numbers on *Abbey Road* run 1 through 17. Asking for track 18
(or using the position directly as a 0-based list index without
adjusting) is out of bounds.


```python
abbey_road_side_a = [
    "Come Together", "Something", "Maxwell's Silver Hammer",
    "Oh! Darling", "Octopus's Garden", "I Want You (She's So Heavy)",
]

track_number = 9  # only 6 tracks are listed above

try:
    # valid indices for a 6-item list are 0-5 -- index 9 is out of bounds,
    # even though "track 9" sounds like a reasonable request
    track_name = abbey_road_side_a[track_number]
    print(f"Track {track_number}: {track_name}")
except IndexError:
    print(f"Track {track_number} is out of range -- only {len(abbey_road_side_a)} tracks listed.")
```

## Mini-challenge: filtering malformed ratings

`raw_ratings` mixes valid numeric strings with malformed entries
(typos, empty strings). Instead of letting one bad entry crash the
whole batch, catch `ValueError` **per item** and skip just that one.


```python
raw_ratings = ["5", "4", "four", "3", "", "5", "2"]  # "four" and "" will each fail int()

valid_ratings = []
skipped_count = 0

for raw_value in raw_ratings:
    try:
        # the try/except sits INSIDE the loop -- so one bad raw_value only
        # skips that one iteration, instead of stopping the whole loop
        valid_ratings.append(int(raw_value))
    except ValueError:
        skipped_count += 1

print("Valid ratings:", valid_ratings)  # valid_ratings is a list of int, malformed entries excluded
print("Skipped (malformed):", skipped_count)
print(f"Average of valid ratings: {sum(valid_ratings) / len(valid_ratings):.2f}")
```

## Let's review

- An unhandled exception crashes the program; `try`/`except` lets it
  keep running instead.
- `ValueError`, `KeyError`, `ZeroDivisionError`, and `IndexError` are
  four of the most common exception types -- each corresponds to a
  different kind of "this data isn't what I expected."
- Catching an exception **inside a loop, per item**, lets you skip
  just the bad entries instead of losing an entire batch of otherwise
  good data.
- `except SomeError:` only catches that specific type -- a different
  exception still propagates and crashes the program. That's often
  what you want: silently swallowing *every* possible error can hide
  real bugs.
