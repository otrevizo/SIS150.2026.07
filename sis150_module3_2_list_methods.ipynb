## Adding elements

`append(x)` adds one item to the end. `extend([...])` adds every item from another list. `insert(i, x)` inserts `x` before position `i`.

```python
groceries = ["milk", "eggs", "bread"]
groceries.append("butter")
groceries
```

```python
groceries.extend(["cheese", "yogurt"])
groceries
```

```python
groceries.insert(1, "apples")  # insert BEFORE position 1
groceries
```

## Removing elements

`remove(x)` deletes the first match by *value*. `pop()` removes and returns the *last* item. `pop(i)` removes and returns the item at position `i`.

```python
groceries.remove("bread")
groceries
```

```python
last_item = groceries.pop()
print(last_item)
print(groceries)
```

```python
first_item = groceries.pop(0)
print(first_item)
print(groceries)
```

## Modifying order

`sort()` and `reverse()` both change the list *in place*.

```python
groceries.sort()
groceries
```

```python
groceries.reverse()
groceries
```

## Getting information

`index(x)` and `count(x)` look things up *without* changing the list.

```python
# reset the list for a fresh lookup
groceries = ["milk", "apples", "eggs", "butter", "cheese"]
groceries.index("milk")
```

```python
groceries = ["milk", "eggs", "milk", "butter", "milk"]
groceries.count("milk")
```

Good practice: prefer list methods like these over manual index tricks -- they make code easier to read.
