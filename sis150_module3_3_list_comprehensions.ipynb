## A list comprehension is a compact for loop

`new_list = [expression for name in iterable]` loops over `iterable`, evaluates `expression` for each item, and returns a brand-new list.

```python
temps_f = [68, 72, 75, 80]
temps_plus10 = [(t + 10) for t in temps_f]
temps_plus10
```

## The same thing, the long way

Compare that one-liner to the equivalent `for` loop.

```python
temps_f = [68, 72, 75, 80]
temps_plus10 = []
for t in temps_f:
    temps_plus10.append(t + 10)
temps_plus10
```

## Converting every element

A comprehension can call a function on each element too.

```python
temps_f = [68, 72, 75, 80]
temps_as_text = [str(t) for t in temps_f]
temps_as_text
```

## Building a list from text tokens

`split()` breaks a string into tokens; a comprehension can convert every token to a number in one line.

```python
raw_scores = "88 92 74 100"
scores = [int(tok) for tok in raw_scores.split()]
scores
```

## Comprehensions over a list of lists

Each `row` below is itself a list -- `sum(row)` adds up one row at a time.

```python
quiz_scores = [[8, 9, 10], [7, 7, 6], [10]]  # one row per student
row_totals = [sum(row) for row in quiz_scores]
row_totals
```

```python
lowest_total = min([sum(row) for row in quiz_scores])
lowest_total
```

## Conditional list comprehensions

Add `if condition` at the end to keep only some elements.

```python
scores = [88, 92, 74, 100, 65, 59]
passing_scores = [s for s in scores if s >= 70]
passing_scores
```

Note: a comprehension always builds a *new* list -- it is not an exact replacement for a `for` loop that modifies a list in place.
