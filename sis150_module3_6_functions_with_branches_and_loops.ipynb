## Branches inside a function

A shipping company charges a different rate depending on package weight. The tiers are just an `if` / `elif` / `else` inside the function body.

```python
def calc_shipping_fee(weight_lbs):
    """Returns the shipping fee for a package given its weight.
    $4.00 flat fee, plus $0.50/lb up to 5 lbs,
    $0.35/lb for the portion from 5.01-20 lbs, and
    $0.20/lb for the portion above 20 lbs."""
    base_fee = 4.00
    if weight_lbs <= 5:
        return base_fee + (weight_lbs * 0.50)
    elif weight_lbs <= 20:
        return base_fee + (5 * 0.50) + ((weight_lbs - 5) * 0.35)
    else:
        return base_fee + (5 * 0.50) + (15 * 0.35) + ((weight_lbs - 20) * 0.20)
```

```python
print(f"3 lb package: ${calc_shipping_fee(3):.2f}")
print(f"12 lb package: ${calc_shipping_fee(12):.2f}")
print(f"40 lb package: ${calc_shipping_fee(40):.2f}")
```

## Loops inside a function, across several functions

Here, one function builds a list, and three more functions each loop over that *same* list to report a different slice of it.

```python
race_times = [5.2, 6.8, 4.9, 7.5, 5.0]  # minutes per mile

def print_all_times(times):
    print("All times:", times)

def print_fast_times(times, threshold=5.0):
    fast = [t for t in times if t <= threshold]
    print("Fast times (<= 5.0):", fast)

def print_slow_times(times, threshold=7.0):
    slow = [t for t in times if t > threshold]
    print("Slow times (> 7.0):", slow)
```

```python
print_all_times(race_times)
print_fast_times(race_times)
print_slow_times(race_times)
```

## A function with multiple return statements

`return` doesn't have to be the last line -- a function can return from different branches. This one "clamps" a grade into a valid range.

```python
def normalize_grade(grade):
    upper_bound = 100
    lower_bound = 50
    if grade >= upper_bound:
        return upper_bound
    elif grade < lower_bound:
        return lower_bound
    else:
        return grade
```

```python
print(normalize_grade(72))
print(normalize_grade(35))
print(normalize_grade(105))
```
